## 1. Import Library
Pustaka yang digunakan:
- `matplotlib.pyplot`: Untuk membuat kanvas grafik, judul, dan label sumbu.
- `seaborn`: Untuk visualisasi statistik tingkat tinggi seperti `regplot` dan `histplot`.
- `pandas`: Untuk membaca file data CSV tabular.


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
```

## 2. Membaca File Data Submission
Membaca file `sample_submission.csv`. Jalur file disesuaikan secara otomatis agar dapat berjalan lancar di lingkungan notebook.


```python
# Menentukan path file csv
csv_path = "sample_submission.csv" if os.path.exists("sample_submission.csv") else r"machine learning competition\sample_submission.csv"
if not os.path.exists(csv_path):
    csv_path = "Submission.csv" if os.path.exists("Submission.csv") else r"D:\file baru\mendata\machine learning competition\sample_submission.csv"

sub_data = pd.read_csv(csv_path)

print(f"Dimensi data: {sub_data.shape}")
print("\n5 Baris Pertama Data Submission:")
display(sub_data.head())
```

## 3. Visualisasi Regresi: House ID vs Sale Price
Kode visualisasi asli dari `visual.py` dirapikan dan dijalankan dalam notebook. 
Plot ini memperlihatkan grafik pencar harga rumah dipetakan berdasarkan ID rumah beserta garis tren regresinya.


```python
# visualization
plt.figure(figsize=(10, 6))
plt.title("Predicted House: Sale Price")
sns.regplot(x=sub_data['Id'], y=sub_data.SalePrice, color='#2b5c8f', scatter_kws={'alpha': 0.6})
plt.xlabel("House ID")
plt.ylabel("House's Price")
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Visualisasi Tambahan: Distribusi & Sebaran Harga Rumah
Untuk melengkapi visualisasi, kita menambahkan analisis distribusi harga rumah (Histogram & Boxplot).


```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# 1. Histogram & KDE
sns.histplot(sub_data['SalePrice'], kde=True, ax=axes[0], color='coral')
axes[0].set_title("Distribusi Harga Rumah Prediksi")
axes[0].set_xlabel("SalePrice ($)")
axes[0].set_ylabel("Frekuensi")
axes[0].axvline(sub_data['SalePrice'].median(), color='red', linestyle='--', label=f"Median: ${sub_data['SalePrice'].median():,.0f}")
axes[0].legend()

# 2. Boxplot
sns.boxplot(x=sub_data['SalePrice'], ax=axes[1], color='salmon')
axes[1].set_title("Boxplot Sebaran Harga Rumah Prediksi")
axes[1].set_xlabel("SalePrice ($)")

plt.tight_layout()
plt.show()

print("=== STATISTIK RINGKAS DATA PREDIKSI ===")
print(sub_data['SalePrice'].describe().round(2))
```
